```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x+3)**2

def gradient(x):
    return 2 * (x + 3)
```

```python
current_x = 2.0
learning_rate = 0.1
iterations = 50
tolerance = 1e-6
```

```python
history_x = [current_x]
history_y = [function(current_x)]
```

```python
print(f"Starting at x = {current_x} and y = {function(current_x)}")
print("-" * 30)

# Gradient Descent Loop
for i in range(iterations):
    grad = gradient(current_x)
    new_x = current_x - grad * learning_rate

    history_x.append(new_x)
    history_y.append(function(new_x))

    if i%5 == 0 or i == iterations-1:
        print(f"Iteration {i+1} : x = {new_x:.4f}, y = {function(new_x):.6f}")

    if abs(new_x - current_x) < tolerance:
        current_x = new_x
        print(f"Convergence at iteration {i+1}")
        break

    current_x = new_x

print("-" * 30)

print(f"Local Minima at x = {current_x:.4f}")
print(f"Minimum Value of y = {function(current_x):.6f}")


# Visualization 
x_range = np.linspace(-10, 5, 100)
y_range = function(x_range)

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, label = "y = (x + 3)^2", color = 'blue')
plt.scatter(history_x, history_y, color = 'red', marker = 'o', s = 50, label = 'Gradient Descent Steps', zorder = 5)
plt.plot(history_x, history_y, color = 'red', linestyle = '--', alpha = 0.6)
plt.title("Gradient Descent to find Local Minima")
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
